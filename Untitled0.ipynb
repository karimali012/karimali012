```python
import pandas as pd
df =pd.read_parquet('yellow_tripdata_2023-01.parquet')
print ("data size is:", df.shape)
df.head()
```

```python
df.isnull().sum()
```

```python
df['total_amount'].describe()
```

```python
df_clean = df[df['total_amount'] > 0]
df_clean = df_clean.dropna()
print("data size after cleaning:",df_clean.shape)
```

```python
df_clean['duration_minutes'] = (df_clean['tpep_dropoff_datetime'] - df_clean['tpep_pickup_datetime']).dt.total_seconds() / 60
df_clean['duration_minutes'].describe()
```

```python
df_final = df_clean[(df_clean['duration_minutes'] >= 1) & (df_clean['duration_minutes'] <= 360)]

df_final = df_final[(df_final['tpep_pickup_datetime'].dt.year == 2023) & (df_final['tpep_pickup_datetime'].dt.month == 1)]

'NYC_Taxi_Cleaned.csv'
df_final.to_csv('NYC_Taxi_Cleaned.csv', index=False)
print("data cleaned and saved successfully! final size:", df_final.shape)
```
